```python
import time

# Scenario: calculate the length (magnitude) of 1 million vectors
# This is doen thousands of times during embedding similarity search

data = np.random.uniform (-1, 1, size= (1_000_000, 3))

# Bad way - Python loop
def magnitude_loop(vectors):
    results = []
    for vec in vectors:
        magnitude =0
        for val in vec:
            magnitude += val ** 2
        results.append(magnitude ** 0.5)
    return results

# Good Way - NumPy vectorized
def magnitude_vectorized(vectors):
    return np.sqrt(np.sum(vectors ** 2, axis=1))

# Compare speed
print("Calculating magnitude of 1 million vetors....\n")

start=time.time()
loop_result = magnitude_loop(data)
loop_time = time.time()-start
print(f"Loop time {loop_time:.4f}s")

start=time.time()
vec_result = magnitude_vectorized(data)
vec_time = time.time() - start
print(f"Vectorized time: {vec_time:.4f}s")

print(f"\nSpeedup: {loop_time/vec_time:.1f}x faster")
print(f"First 5 results match: {np.allclose(loop_result[:5], vec_result[:5])}")
```

```python
# Broadcasting lets you do math between arrays of different shapes
# NumPy stretches the smaller array to match the larger one

# Scenario: normalize all embedding vectors
# Each vector needs to be divided by its own magnitude

np.random.seed(42)
embeddings = np.random.uniform(-1, 1, size=(5,4))
print("Original embeddings:")
print(embeddings)
print(f"Shape: {embeddings.shape}")

# Calculate magnitude of each vector - shape becomes (5,)
magnitudes= np.sqrt(np.sum(embeddings**2, axis=1))
print(f"\n Magnitude shape: {magnitudes.shape}")
print(f"Magnitude: {magnitudes}")

# Reshape for broadcasting - need (5,1) not (5,)
magnitudes_reshaped = magnitudes.reshape(5, 1)
print(f"\n Reshaped magnitude: {magnitudes_reshaped.shape}")

#Divide each row by its magnitude - broadcasting handles the rest 
normalized = embeddings / magnitudes_reshaped
print(f"\n Normalized embeddings:")
print(normalized)

#verify - magnitude of eaach normalized vector should be 1.0
check = np.sqrt(np.sum(normalized**2, axis=1))
print(f"\n Magnitude after normalization( should all be 1.0):")
print(np.round(check, 6))
```

```python
# Cosine similarity measures the angle between two vectors
# Score of 1.0 = identical directions = very similar meaning
# Score of 0.0 = perpendicualr = unrelated
# Score of-1.0 = opposite direction = opposite meaning

import numpy as np

def cosine_similarity(vec1: np.ndarray, vec2: np.ndarray)->float:
    """Calculate similarity between two embedding vectors"""
    dot_product = np.dot(vec1, vec2)
    magnitude1 = np.sqrt(np.sum(vec1**2))
    magnitude2 = np.sqrt(np.sum(vec2**2))
    return dot_product / (magnitude1 * magnitude2)

def cosine_similarity_matrix(query: np.ndarray, docs: np.ndarray)-> np.ndarray:
    """
    Calculate Similarity between one query and ALL documents at once.
    This is what your retriever does on every search.
    """

    #Normalize query
    query_norm = query / np.sqrt(np.sum(query**2))

    # Normalize all docs at once - broadcasting
    doc_magnitudes = np.sqrt(np.sum(docs**2 , axis=1, keepdims=True))
    docs_norm = docs / doc_magnitudes

    #Dot product - query against every doc simultaneously
    return np.dot(docs_norm, query_norm)

# Simulate a real RAG search
np.random.seed(42)

# Fake document chunks and their embeddings
chunks = [
    "RAG combines retrieval and Generation",
    "Vector Databases store embeddings",
    "Python is a programming language",
    "Rag uses semantic search for retrieval",
    "Fine tuning adapts pretrained models"
]

# Simulate embeddings (in real life these come from openAI)
doc_embeddings = np.random.uniform(-1, 1, size=(5,8))

# Simulate query embeddings
query = "how does RAG work?"
query_embedding = np.random.uniform(-1, 1, size=(8,))

# Calculate similarities 
similarities = cosine_similarity_matrix(query_embedding, doc_embeddings)

print("=== RAG Similarity Search ===")
print(f"Query: '{query}'\n")

# Rank by similarity
ranked_indices = np.argsort(similarities)[::-1]

print("Ranked results: ")
for rank , idx in enumerate(ranked_indices):
    print(f"   Rank {rank+1} | Score: {similarities[idx]:.4f} | {chunks[idx]}")

print(f"\nTop result: '{chunks[ranked_indices[0]]}'")
print(f"Score: {similarities[ranked_indices[0]]:.4f}")
```

```python
from typing import List, Tuple

def retrieve_top_k(
        query_embedding: np.ndarray,
        doc_embeddings: np.ndarray,
        chunks: List[str],
        k: int = 3,
        threshold: float = 0.0,
)-> List [Tuple[str, float]]:
    """
    Retrieve top-k most similiar chunks to a query.

    Args:
        query_embedding: embedding vector for the query
        doc_embeddings: matrix of all document embeddings
        chunks: list of original text chunks
        k: number of results to return
        threshold: minimum similarity score to include

        Returns:
        List of (chunk_text, similarity_score) tuples
    """

    # Calculate all similarities at once
    similarities = cosine_similarity_matrix(query_embedding, doc_embeddings)

    #Get top  k indices
    top_k_indices = np.argsort(similarities)[::-1][:k]

    # Builds results - filter by threshold
    results = []
    for idx in top_k_indices:
        score = float(similarities[idx])
        if score >= threshold:
                results.append((chunks[idx], score))

    return results

# Test with different k and threshold values
np.random.seed(42)
doc_embeddings = np.random.uniform(-1, 1, size=(5,8))
query_embedding = np.random.uniform(-1, 1, size=(8,))

chunks = [
     "RAG combines retrieval and generation",
     "Vector Databases store embeddings",
     "Python is a programming language",
     "Fine tuning adapts pretrained models",
     "RAG uses semantic search for retrieval"
]

print("=== Top 3 results(no threshold) ===")
results = retrieve_top_k(query_embedding, doc_embeddings, chunks, k=3)
for chunk, score in results:
     print(f"   [{score:.4f}] {chunk}")

print("=== Top 3 results (threshold = 0.05) ===")
results = retrieve_top_k(query_embedding, doc_embeddings, chunks, k=3, threshold=0.05)
for chunk, score in results:
     print(f" [{score:.4f}] {chunk}")

print("=== Top 5 results (threshold = 0.1) ===")
results = retrieve_top_k(query_embedding, doc_embeddings, chunks, k=5, threshold=0.1)
for chunk, score in results:
     print(f" [{score:.4f}] {chunk}")



```

```python
def get_all_chunks(documents):
    chunks = []
    for doc in documents:
        for chunk in doc:
            chunks.append(chunk)
    return chunks

def get_chunks(documents):
    for doc in documents:
        for chunk in doc:
            yield chunk
```

```python
def read_pdf_chunks(filepath: str, chunk_size: int = 512):
    """
    simulates reading a large PDF and yielding  chunks one at a time.
    """

    fake_pdf_text="word" *10000

    start = 0
    chunk_number= 0

    while start<len(fake_pdf_text):
        chunk=fake_pdf_text[start:start + chunk_size]
        chunk_number += 1
        yield {"chunk_id": chunk_number, "text":chunk, "start":start}
        start += chunk_size

print("Processing PDF chunks:\n")
for chunk in read_pdf_chunks("documents.pdf"):
    if chunk["chunk_id"]<=3:
        print(f"chunk {chunk['chunk_id']}: {chunk['text'][:50]}...")
    if chunk["chunk_id"] == 3:
        break

print("\nGenerator used - PDF never fully loaded into memory!")
```

```python
import time 

def timer(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"{func.__name__}() took {end-start:.6f}s")
        return result
    return wrapper

@timer
def load_document(filepath):
    time.sleep(0.1)
    return f"document loaded from {filepath}"

@timer
def embed_chunks(chunks):
    time.sleep(0.2)
    return f"embedded{len(chunks)} chunks"

@timer
def query_vector(query):
    time.sleep(0.05)
    return f"results for: {query}"

result1=load_document("research.pdf")
result2=embed_chunks(["chunk1", "chunk2", "chunk3"])
result3=query_vector("what is RAG?")

print("\nResults:")
print(result1)
print(result2)
print(result3)
```

```python
import functools

def logger(func):
    @functools.wraps(func)
    def wrapper(* args, **kwargs):
        print(f"[LOG] Calling {func.__name__}")
        print(f"[LOG] Arguments: args={args}, kwargs={kwargs}")
        try:
            result = func(*args, **kwargs)
            print(f"[LOG] {func.__name__} Completed successfully")
            return result
        except Exception as e:
            print(f"[ERROR] {func.__name__} failed: {str(e)}")
            raise
    return wrapper

@logger
def ingest_document(filepath:str, chunk_size:int =512):
    if not filepath.endswith(".pdf"):
        raise ValueError("only PDF files supported")
    return f"ingested {filepath} with chunk_size={chunk_size}"

print("=== Successfull Call ===")
result=ingest_document("research.pdf", chunk_size=256)
print("Result:",result)

print("=== Failed Call ===")
try:
    result = ingest_document("document.txt", chunk_size=256)
except ValueError:
    print("Error was caugth and logged")
    
```

```python
import functools

def cache(func):
    """"Stores result so the same input never gets processed twice"""
    stored_results={}

    @functools.wraps(func)
    def wrapper(*args):
        if args in stored_results:
            print(f"[CACHE HIT] Returning stored result for: {args} ")
            return stored_results[args]
        
        print(f"[CACHE MISS] Processing for first time : {args}")
        result=func(*args)
        stored_results[args]=result
        return result
    return wrapper

@cache
def embed_text(text: str):
    """Simulates an expesive embedding API call"""
    time.sleep(0.5)
    return f"embedding_vector_for_{text[:10]}"

import time

print("=== First calls - nothing cached yet ===")
start = time.time()
r1 = embed_text("What is RAG?")
r2 = embed_text("What is a Vector database?")
print(f"Time taken: {time.time()-start:.2f}s")

print("\n===Second calls - same queries ===")
start = time.time()
r3 = embed_text("what is RAG?")
r4 = embed_text("What is Vector database?")
print(f"Time taken: {time.time()-start:.2f}s")

print("\nResults identical?", r1 == r3, r2 == r4)
```

```python
def cache_normalized(func):
    """cache with text normalization - handles case differences"""
    stored_results = {}

    @functools.wraps(func)
    def wrapper (text : str):
        # Normalize before checking cache
        normalized = text.lower().strip()

        if normalized in stored_results:
            print(f"[CACHE HIT] '{text}'")
            return stored_results[normalized]
        
        print(f"[CACHE MISS]'{text}'")
        result = func(text)
        stored_results[normalized]= result
        return result
    
    return wrapper

@ cache_normalized
def embed_text_v2(text: str):
    time.sleep(0.5)
    return f"embeddings_for_{text.lower().strip()[:10]}"

print("=== Teating normalized cahce ===")
start = time.time()
r1 = embed_text_v2("What is RAG?")
r2 = embed_text_v2("what is RAG?")
r3 = embed_text_v2("WHAT IS RAG?")
r4 = embed_text_v2("what is RAG?")
print(f"\nTime taken :{time.time()-start:.2f}s")
print(f"All results identical? {r1==r2==r3==r4}")
```

```python
class DocumentProcessor:
    """
    Simulates opening and closing a connenction to your vector database.
    In your real RAG project this will be a chromaDB connection.
    """

    def __init__(self, filepath: str):
        self.filepath = filepath
        self.is_open  = False

    def __enter__(self):
        print(f"[OPEN] Opening connection for:{self.filepath}")
        self.is_open = True
        return self
    def __exit__(self, exc_type, exc_val, exc_tb):
        print(f"[CLOSE] Closing connection for: {self.filepath}")
        self.is_open = False

        if exc_type is not None:
            print(f"[ERROR] An error occurred:{exc_val}")

        return False # false = don't suppress the error
    
    def process(self, text: str):
        if not self.is_open:
            raise RuntimeError("Connection not open ")
        return f"processed: {text[:30]}..."
    
# Normal usage
print("=== Normal Usage ===")
with DocumentProcessor("research.pdf") as processor:
    result = processor.process("This is some document text to process")
    print("result:", result)

print("Is open after with block?", processor.is_open)
```

```python
# What happens when an error occurs inside th ewith block?
print("==== Error Case ===")
try:
    with DocumentProcessor("research.pdf") as processor:
        print("Connection is open:", processor.is_open)
        result = processor.process("some text")
        print("Result:", result)

        # Simulate an error mid-processing
        raise ValueError("Corrupt PDF - Cannot read page 3")
    
        # this line never runs
        print("This will never print")

except ValueError as e:
    print(f"Caught error : {e}")

print("\nIs open after error?", processor.is_open)
print("Connection was still closed despite the error!")
```

```python
import time

class VectorDBConnection:
    """
    This is close to what your real ChromaDB connection will look like in week 4 of my project.
    """

    def __init__(self, collection_name: str, user_id: str):
        self.collection_name = collection_name
        self.user_id = user_id
        self.connection = None
        self.query_count = 0

    def __enter__(self):
        print(f"[DB] Connecting to collection: {self.collection_name}")
        print(f"[DB] USer namespace: {self.user_id}")
        # In real code: self.connection = chromadb.client
        self.connection = f"connected_to_{self.collection_name}"
        self.start_time =time.time()
        return self
    
    def __exit__(self, exc_type, exc_val, exc_tb):
        duration = time.time() - self.start_time
        print(f"[DB] Closing Connection")
        print(f"[DB] Total queries made: {self.query_count}")
        print(f"[DB] session Duration: :{duration:.3f}s")
        self.connection= None
        return False
    
    def query(self, text: str):
        self.query_count+=1
        # In real code: return self.connection.query()
        return f"results for '{text}' from {self.collection_name}"
    
print("=== User Query Session ===\n")
with VectorDBConnection ("legal_docs", user_id="jay_123") as db:
        r1 = db.query("what is the penalty clause?")
        print("query 1:", r1)
        
        r2 = db.query("what are the payment terms?")
        print("Query 2:", r2)

        r3 = db.query("who are the parties involved?")
        print("Query 3:", r3)
```

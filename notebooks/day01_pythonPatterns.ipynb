```python
def get_all_chunks(documents):
    chunks = []
    for doc in documents:
        for chunk in doc:
            chunks.append(chunk)
    return chunks

def get_chunks(documents):
    for doc in documents:
        for chunk in doc:
            yield chunk
```

```python
documents=[
    ["chunk1 from doc1", "chunk2 from doc1","chunk3 from doc1" ],
    ["chunk1 from doc2", "chunk2 from doc2"],
    ["chunk1 form doc3", "chunk2 from doc3", "chunk3 from doc3"]
]

all_at_once = get_all_chunks(documents)
print("Type: ", type(all_at_once))
print("All chunks in memory:", all_at_once)

lazy = get_chunks(documents)
print("\nType: ", type(lazy))
print("Generator object: ", lazy)
```

```python
def read_pdf_chunks(filepath: str, chunk_size: int = 512):
    """
    simulates reading a large PDF and yielding  chunks one at a time.
    """

    fake_pdf_text="word" *10000

    start = 0
    chunk_number= 0

    while start<len(fake_pdf_text):
        chunk=fake_pdf_text[start:start + chunk_size]
        chunk_number += 1
        yield {"chunk_id": chunk_number, "text":chunk, "start":start}
        start += chunk_size

print("Processing PDF chunks:\n")
for chunk in read_pdf_chunks("documents.pdf"):
    if chunk["chunk_id"]<=3:
        print(f"chunk {chunk['chunk_id']}: {chunk['text'][:50]}...")
    if chunk["chunk_id"] == 3:
        break

print("\nGenerator used - PDF never fully loaded into memory!")
```
